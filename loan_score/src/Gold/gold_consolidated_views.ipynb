```python
spark.sql('''
create or replace view vw_customers_loan_defaulter AS
SELECT c.member_id, c.emp_title, c.employment_length, c.home_ownership, c.annual_income, c.address_state, c.zip_code, c.country,c.grade, c.sub_grade, 
c.verification_status, c.total_high_credit_limit, c.application_type,c.annual_income_joint, c.verification_status_joint, l.loan_id, l.Loan_amount, l.funded_amount, 
l.loan_term, l.loan_years,l.int_rate, l.installment,l.issue_date, l.loan_status, l.loan_purpose, l.title, lr.total_rec_principal, lr.total_rec_interest, 
lr.total_rec_late_fee, lr.total_payment, lr.last_payment_amount, lr.last_payment_date, lr.next_payment_date, ldd.delinq_2yrs,ldd.delinq_amount,
ldd.months_since_last_delinq, ldp.public_rec, ldp.public_bankruptcies,ldp.inquiry_6_months, ldp.months_since_last_public_record
FROM silver_enriched.customers c
LEFT JOIN silver_cleaned.loans l on c.member_id = l.member_id
LEFT JOIN silver_cleaned.loans_repayment lr on l.loan_id = lr.loan_id
LEFT JOIN silver_enriched.loans_defaulter_delinq ldd on c.member_id = ldd.member_id
LEFT JOIN silver_enriched.loans_defaulter_publicrecord ldp on c.member_id = ldp.member_id
''')
```

```python
display(spark.sql('select * from vw_customers_loan_defaulter'))
```
